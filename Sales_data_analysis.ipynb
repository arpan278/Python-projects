```python
#Question 5. Calculate the total revenue, total cost and total profit generated by the business?

total_revenue = df["Revenue"].sum()
total_cost = df["Cost"].sum()
total_profit = df["Profit"].sum()

print("Total Revenue: Rs.", format(total_revenue, ",.2f"))
print("Total Cost: Rs.", format(total_cost, ",.2f"))
print("Total Profit: Rs.", format(total_profit, ",.2f"))
```

```python
#Question 6. Calculate the average order value and overall profit margin?

total_orders = df["Order_ID"].nunique()

average_order_value = total_revenue / total_orders

profit_margin = (total_profit / total_revenue) * 100

print("Total Orders:", total_orders)

print("Average Order Value: Rs.",
      round(average_order_value, 2))

print("Profit Margin:",
      round(profit_margin, 2), "%")
```

```python
#Question 7. Find the total quantity sold and revenue generated by each product?

product_analysis = df.groupby("Product").agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Total_Profit=("Profit", "sum")
)

print("Product-wise Sales Analysis:")

display(product_analysis.sort_values(
    by="Total_Revenue",
    ascending=False
))


```

```python
#Question 8. Identify the top five products based on total sales revenue?

top_products = df.groupby("Product")["Revenue"].sum()

top_5 = top_products.sort_values(
    ascending=False
).head(5)

print("Top Five Products by Revenue:")

display(top_5.to_frame())
```

```python
#Question 9. Analyse the sales performance of different regions and identify the best-performing region?

region_analysis = df.groupby("Region").agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order_ID", "count")
)

print("Region-wise Sales Analysis:")

display(region_analysis.sort_values(
    by="Total_Revenue",
    ascending=False
))

best_region = region_analysis["Total_Revenue"].idxmax()

print("Best Performing Region:", best_region)
```

```python
#Question 11. Analyse customer payment preferences and identify the most frequently used payment method?

payment_analysis = df["Payment_Mode"].value_counts()

print("Payment Method Analysis:")

display(payment_analysis.to_frame(
    name="Number_of_Orders"
))

print("Most Popular Payment Method:",
      payment_analysis.idxmax())
```
